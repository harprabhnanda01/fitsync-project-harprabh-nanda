````python
# --- Student TODO: implement only this function. Keep it minimal. ---
from openai import OpenAI

def call_openai(prompt: str) -> str:
    """
    Return the plain text output from the selected OpenAI model through OpenRouter.
    Use OPENROUTER_API_KEY, OPENROUTER_BASE_URL, and OPENAI_MODEL.
    """
    client = OpenAI(
        api_key=OPENROUTER_API_KEY,
        base_url=OPENROUTER_BASE_URL,
    )
    response = client.chat.completions.create(
        model=OPENAI_MODEL,
        messages=[{"role": "user", "content": prompt}],
        temperature=0.3,
        max_tokens=250,
    )
    return response.choices[0].message.content or ""
````

# Reflection – Multibot Chat Playground

## Introduction

The Multibot Chat Playground assignment helped me understand how multiple Large Language Models (LLMs) can be integrated into a single application using OpenRouter. The objective was to send identical prompts to different models, collect their responses, and compare their performance across different tasks. For this experiment, I used OpenAI GPT-4o-mini and Google Gemini 2.5 Flash. Both models were accessed through a single OpenRouter API key, making the integration process simpler and more convenient.

## 1. Comparison of Model Performance

I tested both models using three tasks: creative writing, analytical reasoning, and instruction following.

**Creative Writing:** Both models were asked to generate a four-line poem about learning by testing code. GPT-4o-mini produced a poem focusing on learning through errors, testing, and trial and error. Gemini 2.5 Flash created a poem with a more rhythmic and straightforward style, using phrases such as "bring the light" and "a lesson learned." Both responses followed the four-line requirement and remained relevant to the topic. GPT presented a slightly more reflective tone, while Gemini used simpler and more catchy language. Both performed well in this task.

**Analytical Reasoning:** Both models were asked to summarize the advantages and disadvantages of using multiple LLM providers in one sentence. GPT explained that multiple providers improve creativity, formatting reliability, and speed comparison but introduce integration and testing challenges. Gemini communicated almost the same meaning using fewer words and a concise structure. Both responses were accurate, relevant, and followed the one-sentence requirement of fewer than 40 words. Gemini's response was slightly more concise, while GPT's wording was more descriptive.

**Instruction Following:** Both models were instructed to return only valid JSON containing exactly two to-do items, with a task string and a done boolean. GPT-4o-mini successfully returned plain JSON with the required structure. Gemini also generated the correct two items and data types, but wrapped its response in Markdown JSON code fences. This meant that its complete response did not strictly satisfy the instruction to return only valid JSON. This comparison showed that GPT performed better in strict formatting compliance.

## 2. Differences in Speed, Formatting, Tone and Creativity

The experiment demonstrated that different models can produce different styles of responses even when they receive exactly the same prompt.

In creative writing, both models generated meaningful poems, but their writing styles differed. GPT used a reflective approach, whereas Gemini produced a more rhythmic and simple poem. In analytical reasoning, both models maintained clarity and accuracy, although Gemini used a more concise sentence.

The biggest difference appeared in instruction following. GPT returned clean JSON without additional formatting, while Gemini included Markdown code fences. This small difference can become important when an application expects machine-readable output.

Response speed is another important factor when comparing LLMs. However, I did not record exact response times during this experiment, so I cannot conclude which model was faster based on measured data.

## 3. Role of OpenRouter and API Mechanics

OpenRouter played an important role by providing a common API interface for accessing different models. Instead of managing separate API keys for different providers, I used a single OpenRouter API key.

The notebook used the OpenAI-compatible API endpoint and the OpenAI Python SDK. Model selection variables allowed me to route requests to GPT-4o-mini and Gemini 2.5 Flash using separate functions.

Important API parameters included the model name, prompt, temperature, and maximum output tokens. Temperature controls the randomness of generated responses, while maximum tokens help control response length and API usage. Using identical prompts made the comparison more meaningful because the models received the same instructions.

## 4. Designing an Automatic Model Routing System

An automatic model routing system could improve the efficiency of a multi-model application. Instead of sending every request to the same model, the application could identify the task type and select a suitable model.

For example, a creative writing request could be routed to a model that generates engaging content. Analytical questions could be assigned to a model that produces concise and accurate answers. Tasks requiring structured JSON could be routed to a model that follows formatting instructions reliably.

The routing system could also consider API cost, response time, token usage, and model availability. Additional validation could be used to check whether the generated JSON follows the required schema before passing it to another application.

## 5. Strengths and Limitations of Multiple LLM Providers

Using multiple LLM providers offers flexibility and allows developers to compare different models for different tasks. It also reduces dependence on a single provider and makes it possible to select models according to application requirements.

However, multiple models can produce inconsistent outputs and interpret instructions differently. Integration also introduces additional testing and maintenance requirements. API availability, pricing, rate limits, and authentication errors can affect application reliability.

The JSON comparison particularly demonstrated why model outputs should be validated instead of blindly trusted.

## Conclusion

This assignment provided practical experience in building a multi-model AI application using Python and OpenRouter. I learned how to configure API access, implement model-calling functions, send identical prompts to different models, and compare their responses.

The experiment showed that GPT-4o-mini performed better in strict JSON formatting, while both models performed well in creative writing and analytical reasoning. The most important learning was that no single model should automatically be considered the best for every task. Model selection should depend on quality, accuracy, speed, formatting requirements, and cost. Overall, the Multibot Chat Playground demonstrated how multiple LLM providers can make AI applications more flexible and adaptable.
